```python
df = df.replace(1, 'Yes')
df = df.replace(0, 'No')
```

```python
df['heart disease'].value_counts()
```

```python
df.corr(numeric_only=True)
```

```python
df['heart disease'].value_counts().plot(kind='bar',figsize=(8,5))
plt.title('Heart Disease Distribution')
plt.xlabel('Heart Disease')
plt.ylabel('Number of Patients')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=15, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['cholestrol'], kde=True)
plt.title('Cholesterol Distribution')
plt.xlabel('Cholesterol')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['BP'], kde=True)
plt.title('Blood Pressure Distribution')
plt.xlabel('Blood Pressure')
plt.ylabel('Frequency')
plt.show()
```

```python
df.groupby('heart disease')['cholestrol'].mean().plot(kind='bar',figsize=(8,5))
plt.title('Average Cholesterol by Heart Disease')
plt.xlabel('Heart Disease')
plt.ylabel('Average Cholesterol')
plt.show()
```

```python
df.groupby('heart disease')['age'].mean().plot(kind='bar',figsize=(8,5))
plt.title('Average Age by Heart Disease')
plt.xlabel('Heart Disease')
plt.ylabel('Average Age')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(),annot=True)
plt.title('Heart Disease Correlation Heatmap')
plt.show()
```
